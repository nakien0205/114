```python

from pathlib import Path
import hashlib, json, os, shutil, zipfile

EXPECTED_MANIFEST_SHA256 = 'dfe80c62a2553405b54ea19f8f548d654c5b1118bc6f56a4cfb2f596f410c806'
EXPECTED_ZIP_SHA256 = '16f18284ccfa7cb868496b5f098aef2fca46de147e7fbefb83152fa1466967c3'
BUNDLE_FILENAME = 'fire_training_shared_v1.zip'
MANIFEST_FILENAME = 'shared-assets-manifest.json'

def bundle_hash(path):
    digest = hashlib.sha256()
    with Path(path).open('rb') as stream:
        for block in iter(lambda: stream.read(1024*1024), b''):
            digest.update(block)
    return digest.hexdigest()

def locate_bundle(search_root, explicit=None):
    if explicit:
        path = Path(explicit)
        if not path.exists():
            raise FileNotFoundError('ASSET_ROOT does not exist: '+str(path))
        return path
    candidates = []
    visited = set()
    for current, directories, files in os.walk(search_root, followlinks=True):
        folder = Path(current)
        if folder.resolve() in visited:
            directories[:] = []
            continue
        visited.add(folder.resolve())
        if MANIFEST_FILENAME in files:
            if bundle_hash(folder/MANIFEST_FILENAME)==EXPECTED_MANIFEST_SHA256:
                candidates.append(folder)
            directories[:] = []
            continue
        if BUNDLE_FILENAME in files:
            path = folder/BUNDLE_FILENAME
            if bundle_hash(path)==EXPECTED_ZIP_SHA256:
                candidates.append(path)
        directories[:] = [name for name in directories if name.lower() not in
            {'images','labels','test','testing','.git','__pycache__'}]
    if not candidates:
        raise RuntimeError('Attach a Kaggle Dataset created from fire_training_shared_v1.zip. '
            'Both the ZIP file and Kaggle-auto-extracted folders are supported. '
            'Alternatively set ASSET_ROOT to the exact ZIP/folder. No matching verified bundle was found.')
    # Duplicate copies have the same immutable identity; prefer an extracted copy.
    return sorted(candidates, key=lambda p:(not p.is_dir(),str(p)))[0]

def materialize_bundle(source, destination):
    source, destination = Path(source), Path(destination)
    if source.is_file():
        if bundle_hash(source)!=EXPECTED_ZIP_SHA256:
            raise RuntimeError('Companion ZIP hash mismatch; use the ZIP supplied with this notebook.')
        destination.mkdir(parents=True, exist_ok=False)
        with zipfile.ZipFile(source) as archive:
            if len(archive.namelist())!=len(set(archive.namelist())):
                raise RuntimeError('Duplicate bundle archive entries')
            for name in archive.namelist():
                if not (destination/name).resolve().is_relative_to(destination.resolve()):
                    raise RuntimeError('Invalid companion archive path')
            archive.extractall(destination)
    elif source.is_dir():
        if bundle_hash(source/MANIFEST_FILENAME)!=EXPECTED_MANIFEST_SHA256:
            raise RuntimeError('Companion folder manifest mismatch')
        manifest = json.loads((source/MANIFEST_FILENAME).read_text())
        destination.mkdir(parents=True, exist_ok=False)
        for name, expected in manifest['files'].items():
            target = destination/name
            if not target.resolve().is_relative_to(destination.resolve()):
                raise RuntimeError('Invalid companion manifest path')
            if bundle_hash(source/name)!=expected:
                raise RuntimeError('Companion file hash mismatch: '+name)
            target.parent.mkdir(parents=True, exist_ok=True)
            shutil.copy2(source/name,target)
        shutil.copy2(source/MANIFEST_FILENAME,destination/MANIFEST_FILENAME)
    else:
        raise RuntimeError('ASSET_ROOT must be a ZIP file or extracted folder')
    if bundle_hash(destination/MANIFEST_FILENAME)!=EXPECTED_MANIFEST_SHA256:
        raise RuntimeError('Companion manifest identity mismatch')
    manifest=json.loads((destination/MANIFEST_FILENAME).read_text())
    for name, expected in manifest['files'].items():
        if bundle_hash(destination/name)!=expected:
            raise RuntimeError('Extracted companion file hash mismatch: '+name)
    return destination

# Load attached assets. This cell does not import Torch or start training.
BUNDLE = locate_bundle(Path('/kaggle/input'), ASSET_ROOT)
ASSETS = materialize_bundle(BUNDLE, RUN_ROOT/'shared-assets')
print('Verified companion assets:', BUNDLE)


# Model/runner code travels with this notebook, not with the reusable upload.
import base64, io
SHARED_ASSETS = ASSETS
ASSETS = RUN_ROOT/'assets'
ASSETS.mkdir(parents=True,exist_ok=False)
for source in SHARED_ASSETS.iterdir():
    if source.is_file():
        target = ASSETS/source.name
        try:
            target.symlink_to(source.resolve())
        except OSError:
            shutil.copy2(source,target)
EMBEDDED_CODE_MANIFEST = {'shared_manifest_sha256': 'dfe80c62a2553405b54ea19f8f548d654c5b1118bc6f56a4cfb2f596f410c806', 'experiment_zip_sha256': '41d9d24e6ee511df4545b07675198a91c7f2deb21f8fccaa1f517f55f77cb547', 'files': {'README.md': 'b89a4d54d6f7e30d03ce8ddcfbdb3b1af342847f0426fdba9597d1dbd371a012', 'baseline_models.py': '9434b66b42a70e17da6451f6bc66e07e2ec127858972d9bfbcca0a9e18993269', 'data_runtime.py': '0032d2d8c88c6f339bdd9b0161ea2d897935af8cdfe5cabcb95b9eac95c913cc', 'experiment-spec.json': '99eea942387e34e309646813cb5976286b5ea5d9e7ca147fb627a703e9a91587', 'operating_points.py': '17ee1eac1aabf8052de16da27116d71c061907b5cae611773521d8161d45b384', 'protocol.json': '2c8176122d40293828ae67e5b020caf2ed3b54b60a2d24a103f476628a64b24d', 'runner.py': '257117bff11fa3e6d7e82c11bdf620ef39bf46571603949352cd9a9f84310b3d', 'size_evaluator.py': '6bbfa9bb85f5889526c17a53c17840fa302e70c8f3585177db288c00b3dd1bc8', 'src_attention.py': 'd14cc3a174517e705cb6a2e4cfb1be7fe0d87f797fe7ae7c5fab34f9716d3518', 'src_models.py': '2096dd2898e921ecd84006320aca06b67318c907ca9a0c35fd64df3fab354d67', 'suite.py': '418189aa47da752d094365756e5838c33a48bc1a88fe3fa9a7074fd98daa735b'}}
payload = base64.b64decode('')
assert hashlib.sha256(payload).hexdigest()==EMBEDDED_CODE_MANIFEST['experiment_zip_sha256']
with zipfile.ZipFile(io.BytesIO(payload)) as archive:
    assert set(archive.namelist())==set(EMBEDDED_CODE_MANIFEST['files'])
    for name,expected in EMBEDDED_CODE_MANIFEST['files'].items():
        destination = ASSETS/name
        assert destination.resolve().is_relative_to(ASSETS.resolve())
        assert not destination.exists(), 'Shared assets are immutable; notebook code must use a separate filename.'
        raw = archive.read(name)
        assert hashlib.sha256(raw).hexdigest()==expected
        destination.parent.mkdir(parents=True,exist_ok=True)
        destination.write_bytes(raw)
(ASSETS/'embedded-code-manifest.json').write_text(json.dumps(EMBEDDED_CODE_MANIFEST,indent=2))
print('Shared companion verified; notebook-specific code restored:',len(EMBEDDED_CODE_MANIFEST['files']),'files')
```
